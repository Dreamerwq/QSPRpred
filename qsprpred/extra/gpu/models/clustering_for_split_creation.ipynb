# `clustering_for_split_creation.ipynb`

## Overview
This notebook automates the creation of 10-fold clustered data splits (Train, Validation, Test) for three molecular datasets: **A2AR**, **CK1**, and **hERG**. It uses the `qsprpred` library to apply Graph-Based Molecular Topology (`GBMTDataSplit`) to ensure chemically distinct splits.

## Workflow 
The script executes 10 iterations (varying random states 42-51) for each of the three datasets:
1. **Feature Calculation:** Computes Morgan Fingerprints (radius=3, nBits=2048) for the molecules.
2. **Data Splitting:** * Extracts a 20% independent **Test** set.
   * Splits the remaining data (using a 0.25 fraction) to generate a 20% **Validation** set and a 60% **Train** set.
3. **Filtering:** Isolates the `Drug` (SMILES) and `Y` (Target) columns.
4. **Exporting:** Logs the dataset shapes and saves the splits.

## Outputs
Files are exported to `Bachelors_Thesis_Experimental_Data/<Dataset_Name>/data_splits_clustered/` using the following naming conventions (1-10):
* `<dataset>_train_<i>`
* `<dataset>_val_<i>`
* `<dataset>_test_<i>`

```python
import os
import qsprpred

from qsprpred.data import QSPRDataset, RandomSplit
from qsprpred.data.descriptors.fingerprints import MorganFP
import pandas as pd
from qsprpred.data.sampling.splits import GBMTDataSplit
```

## A2AR

```python
for i in range(1,11):
    dataset = QSPRDataset.fromTableFile(
    filename="data/src/datasets/a2ar/data/A2AR.csv",
    store_dir="dataset_outputs/a2ar/data",
    name="A2ARDataset",
    target_props=[{"name": "Y", "task": "SINGLECLASS", "th": [0.5]}],
    random_state=41+i,
    smiles_col = 'Drug',
    sep=','
    )
    dataset.prepareDataset(
    split=GBMTDataSplit(test_fraction=0.2, dataset=dataset),
    feature_calculators=[MorganFP(radius=3, nBits=2048)],
    recalculate_features=True,
    )
    dataset2 = QSPRDataset.fromTableFile(
    filename="data/src/datasets/a2ar/data/A2AR.csv",
    store_dir="dataset_outputs/a2ar/data",
    name="A2ARDataset",
    target_props=[{"name": "Y", "task": "SINGLECLASS", "th": [0.5]}],
    random_state=41+i,
    smiles_col = 'Drug',
    sep=','
    )
    dataset2.df = dataset.getDF().loc[~dataset.getDF().index.isin(dataset.X_ind.index)].copy()
    dataset2.prepareDataset(
    split=GBMTDataSplit(test_fraction=0.25, dataset=dataset2),
    feature_calculators=[MorganFP(radius=3, nBits=2048)],
    recalculate_features=True,
    )
    
    df_test = dataset.getDF().loc[dataset.getDF().index.isin(dataset.X_ind.index)][["Y", "Drug"]]
    df_train = dataset2.getDF().loc[~dataset2.getDF().index.isin(dataset2.X_ind.index)][["Y", "Drug"]]
    df_val = dataset2.getDF().loc[dataset2.getDF().index.isin(dataset2.X_ind.index)][["Y", "Drug"]]

    print(i)
    print(df_train.shape)
    print(df_val.shape)
    print(df_test.shape)
    df_train.to_csv(f"Bachelors_Thesis_Experimental_Data/A2AR/data_splits_clustered/a2ar_train_{i}")
    df_val.to_csv(f"Bachelors_Thesis_Experimental_Data/A2AR/data_splits_clustered/a2ar_val_{i}")
    df_test.to_csv(f"Bachelors_Thesis_Experimental_Data/A2AR/data_splits_clustered/a2ar_test_{i}")
```
